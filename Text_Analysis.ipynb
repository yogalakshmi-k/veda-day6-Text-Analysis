```python

import re
from collections import Counter

text = """
Artificial intelligence helps computers learn from data.
Text analysis is an important part of natural language processing.
Artificial intelligence can analyze text and identify repeated words.
"""

# 1. Count words (ignore punctuation and capitalization)
words = re.findall(r"\b[a-zA-Z]+\b", text.lower())
word_count = len(words)

# 2. Count characters (including spaces and punctuation)
character_count = len(text)

# 3. Count vowels
vowels = sum(1 for char in text.lower() if char in "aeiou")

# 4. Identify repeated words
word_frequency = Counter(words)
repeated_words = {
    word: count
    for word, count in word_frequency.items()
    if count > 1
}

# Display results
print("TEXT ANALYSIS REPORT")
print("-" * 30)
print("Total words:", word_count)
print("Total characters:", character_count)
print("Total vowels:", vowels)
print("Repeated words:", repeated_words)
```
